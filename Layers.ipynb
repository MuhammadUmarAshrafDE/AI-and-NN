```python
import numpy as np
input = [[1.3,2.4,3.6,-1.8],
         [0.5,-0.84,-0.53,-0.3],
         [-0.21,-0.62,0.16,0.78],
         [0.14,-0.45,0.62,-0.19]]

weights1=  [[0.2,0.8,-0.5,0.3],
         [0.5,-0.84,0.53,-0.3],
        [-0.21,-0.62,0.16,0.78],
        [0.14,-0.45,0.62,-0.19]]

bias1    =   [2,4,0.8,-1.2]

weights2=  [[0.2,0.8,-0.5,0.3],
         [0.5,-0.84,0.53,-0.3],
        [-0.21,-0.62,0.16,0.78],
        [0.14,-0.45,0.62,-0.19]]

bias2    =   [2,4,0.8,-1.2]

output1 = np.dot(input, np.array(weights1).T) +bias1

#apply ReLU activation function
output1 = np.maximum(0,output1)
print("Output for the first layer:", *output1, sep=',')

output2 = np.dot(output1, np.array(weights2).T) +bias2
#apply ReLU activation function
output2 = np.maximum(0, output2)
```
