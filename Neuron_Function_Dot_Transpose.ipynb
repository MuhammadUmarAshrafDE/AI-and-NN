```python
import numpy as np
input = [[3.0,2.6,1.7,2.1],
         [0.4,0.2,-0.33,-0.22],
         [-0.77,-0.54,-0.23,0.7]]

weights=  [[1.0,2.0,1.0,2.0],
         [0.4,0.3,-0.5,-0.1],
        [-0.66,-0.5,-0.3,0.7]]

bias    =   [2.0,1.1,0.5]
#Perform matrix multiplication using np.dot
#convert weights to a Numpy array and then transpose it using .T

outputs = np.dot(input, np.array(weights).T) +bias

print(outputs)
```
